## Problem 2) LSST Data Products

Rubin produces multiple photometric measurements for every source:

$~~~~$image subtraction (science image $-$ template) photometry

$~~~~$direct photometry (on individual, calibrated science images)

**`DiaSource`**

When a source varies by $\ge 5\sigma$ relative to a reference, it becomes a `DiaSource`, with a flux measured by the change in brightness relative to the template.

(This can be hard for stellar/AGN light curves)

**`DiaObject`**

Multiple `DiaSource`s at the same position on the sky are aggregated to create a `DiaObject`.

*Note* – a `DiaSource` light curve is limited to epochs when the flux varies by $\ge 5\sigma$, meaning small changes in the flux are missing producing a biased light curve.

`DiaForcedSource` is "forced" photometry at the position of a `DiaObject` on every image that covers the location of that source.

**`ForcedSource`**

For all `Object`s detected in deep coadds, forced photometry is measured on every individual image.

Here's the upshot: use `DiaSource` to find transients, and `ForcedSource` to study persistent sources. 

**Real-time alerts**

Within $\sim$$60~\mathrm{s}$ of each visit, the difference imaging pipeline is run and up to $\sim$$10$ million new `DiaSource`s are found per night.

Alerts contain the `DiaSource`, its `DiaObject`, up to 12 months of previous detections and forced photometry, and image cutouts (science, template, difference).

The full alert stream goes to seven community *brokers*: ALeRCE, AMPEL, ANTARES, Babamul, Fink, Lasair, and Pitt-Google.

Brokers:

$~~~~$cross-match alerts with other catalogs

$~~~~$compute light-curve features

$~~~~$classify sources with machine learning

$~~~~$let users filter the stream down to the sources relevant to their science

Periodicity is the most fundamental signal in astronomical light curves. Periodic signals are always very closely related to fundamental physics (orbits, pulsations, rotation). If you can measure the period of a source, you can know something fundamental about its nature.

## Problem 3) Beyond Periodicity

AGN, young stellar objects, blazars, cataclysmic variables, and stellar flares all vary ... without periodicity.

To characterize stochastic variables, we can treat a light curve as a noisy realization of a random process: 

$$y_i = x(t_i) + \epsilon_i, \qquad \epsilon_i \sim \mathcal{N}(0, \sigma_i^2)$$

where $f(t)$ is the flux as a function of time and $\sigma_i$ are the (heteroskedastic) measurement uncertainties.

A process is (weakly) stationary if its mean and variance do not change with time, and the covariance between two observations depends only on their separation, $\tau$: 

$$\gamma(\tau) = \mathrm{Cov}[f(t), f(t+\tau)]$$

Normalizing by the variance, $\gamma(0) = \sigma^2$, gives the autocorrelation function (ACF): 

$$\rho(\tau) = \frac{\gamma(\tau)}{\gamma(0)}, \qquad \rho(0) = 1$$

It is not safe to assume all sources are stationary, but AGN, YSOs, and variable stars can be treated as stationary when observed over long baselines.

The ACF measures how similar a light curve is to a time-shifted copy of itself. 

For $N$ evenly spaced observations, the ACF at lag $k$ is estimated as: 

$$\hat\rho(k) = \frac{\sum_{i=1}^{N-k}(y_i-\bar y)(y_{i+k}-\bar y)}{\sum_{i=1}^{N}(y_i-\bar y)^2}$$

Reading an ACF:

$~~~~$$\hat\rho(k) \approx 0$ for all $k > 0$ $\longrightarrow$ uncorrelated (white) noise

$~~~~$$\hat\rho(k)$ decays smoothly $\longrightarrow$ correlated (red) noise, the decay sets a characteristic timescale

$~~~~$$\hat\rho(k)$ oscillates $\longrightarrow$ periodic signal

<img style="display: block; margin-left: auto; margin-right: auto; width: 70%" src="./images/acf_examples.png" align="middle">

Our first definition of the ACF requires regular sampling (a pair separated by $k$ samples is separated by $k \Delta t$ in time). 

LSST will produce irregular sampling. 

Any simple interpolation onto a regular grid introduces correlations that biases the ACF.

The Discrete Correlation Function (DCF) instead uses every pair of observations ($i \ne j$), and bins by their separations $\Delta t_{ij}$: 

$$\mathrm{UDCF}_{ij} = \frac{(y_i - \bar y)(y_j - \bar y)}{s^2 - \bar\sigma^2}$$

$$\mathrm{DCF}(\tau) = \frac{1}{M}\sum \mathrm{UDCF}_{ij}, \quad \mathrm{for~the~} M \mathrm{~pairs~with~} |\Delta t_{ij} - \tau| < \Delta\tau/2$$

where $s^2$ is the sample variance and $\bar\sigma^2$ is the mean measurement variance (which removes the contribution of the noise).

<div align="right"> <font size="-3">(Edelson & Krolik 1988) </font></div>

If the same approach is applied to two different light curves, that produces the cross-correlation function. 

This is useful for AGN reverberation mapping and gravitational stron lensing time delays. 

The structure function (SF) measures the typical difference between two observations separated by $\tau$: 

$$SF^2(\tau) = \langle [y(t+\tau) - y(t)]^2 \rangle$$

$~~~~$compute $(y_j - y_i)^2$ for every pair of observations

$~~~~$bin by $\Delta t_{ij}$ and average

$~~~~$there are no assumptions about the time between observations so irregular sampling is not a problem

For a stationary process, the SF and the ACF are directly related: 

$$SF^2(\tau) = 2\sigma^2[1 - \rho(\tau)] + 2\sigma_n^2$$

where $\sigma^2$ is the variance of the signal and $\sigma_n^2$ is the noise variance.

Three regimes:

$~~~~$short $\tau$ $\longrightarrow$ noise floor ($2\sigma_n^2$)

$~~~~$intermediate $\tau$ $\longrightarrow$ the SF rises

$~~~~$long $\tau$ ($\rho \rightarrow 0$) $\longrightarrow$ plateau at $2\sigma^2 + 2\sigma_n^2$

<img style="display: block; margin-left: auto; margin-right: auto; width: 70%" src="./images/structure_function.png" align="middle">

$\tau$ can only be reliably recovered when the baseline is much longer than $\tau$.

The SF is widely used to study quasar variability.

Quasar SFs rise as a power law and flatten on timescales of hundreds of days (MacLeod et al. 2010). The SF parameters alone can separate quasars from stars (Schmidt et al. 2010).

(as a quick preview)

It is possible to model the covariance between observations directly, as we will do later this week...

Gaussian processes specify the covariance with a kernel, e.g., the damped random walk: 

$$k(\tau) = \sigma^2 e^{-|\tau|/\tau_d}$$

The likelihood of irregularly sampled data with heteroskedastic uncertainties can then be computed exactly, and quickly.

<div align="right"> <font size="-3">(Foreman-Mackey et al. 2017; Foreman-Mackey 2018) </font></div>

## Appendix: Code for Figures

The code below generates `images/acf_examples.png` and `images/structure_function.png`.

Both figures use simulated data: a damped random walk (DRW), white noise, and a sinusoid.

```python
import os
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({"font.size": 12,
                     "axes.spines.top": False,
                     "axes.spines.right": False})
os.makedirs("images", exist_ok=True)


def simulate_drw(t, tau, sigma, rng):
    """Damped random walk (Ornstein-Uhlenbeck process) evaluated at times t.

    Uses the exact update for arbitrary (irregular) time steps.
    tau = damping timescale, sigma = long-term standard deviation.
    """
    x = np.empty(len(t))
    x[0] = rng.normal(0, sigma)
    for i in range(1, len(t)):
        a = np.exp(-(t[i] - t[i-1]) / tau)
        x[i] = a * x[i-1] + rng.normal(0, sigma * np.sqrt(1 - a**2))
    return x


def acf_regular(y, kmax):
    """ACF estimator for evenly spaced observations, lags k = 0, ..., kmax-1."""
    y = y - y.mean()
    denom = np.sum(y**2)
    return np.array([np.sum(y[:len(y)-k] * y[k:]) / denom for k in range(kmax)])
```

```python
# Figure: example light curves and their ACFs (regular sampling, dt = 1 d)
rng = np.random.default_rng(42)
N = 1000
t = np.arange(N, dtype=float)

lcs = [("White noise",
        rng.normal(0, 1, N)),
       (r"Correlated noise (DRW, $\tau = 20$ d)",
        simulate_drw(t, 20, 1, rng) + rng.normal(0, 0.2, N)),
       (r"Periodic ($P = 50$ d) + noise",
        np.sin(2*np.pi*t/50) + rng.normal(0, 0.5, N))]

kmax = 150
fig, ax = plt.subplots(3, 2, figsize=(11, 7.5),
                       gridspec_kw={"width_ratios": [1.6, 1]})
for i, (label, y) in enumerate(lcs):
    ax[i, 0].plot(t[:400], y[:400], "o", ms=2.5, color="#2a6f97")
    ax[i, 0].set_ylabel("flux")
    ax[i, 0].set_title(label, loc="left", fontsize=12)
    ax[i, 1].axhline(0, color="0.6", lw=1)
    ax[i, 1].plot(np.arange(kmax), acf_regular(y, kmax), color="#c8553d", lw=2)
    ax[i, 1].set_ylim(-1.05, 1.05)
    ax[i, 1].set_ylabel(r"$\hat\rho(k)$")
ax[0, 1].set_title("ACF", loc="left", fontsize=12)
ax[2, 0].set_xlabel("time (d)")
ax[2, 1].set_xlabel("lag $k$ (d)")
fig.tight_layout()
fig.savefig("images/acf_examples.png", dpi=150)
```

```python
# Figure: structure function of a DRW observed with seasonal, irregular sampling
rng = np.random.default_rng(2021)
tau, sigma, sigma_n = 100., 0.2, 0.05     # DRW timescale (d), signal rms, noise rms (mag)

# 10 seasons of ~200 d, 80 randomly timed observations per season
t = np.concatenate([365.25*yr + np.sort(rng.uniform(0, 200, 80)) for yr in range(10)])
y = simulate_drw(t, tau, sigma, rng) + rng.normal(0, sigma_n, len(t))

# all pairs (i < j), binned in log(dt)
i, j = np.triu_indices(len(t), 1)
dt = np.abs(t[j] - t[i])
dm2 = (y[j] - y[i])**2
bins = np.logspace(-1.3, np.log10(3650), 30)
idx = np.digitize(dt, bins)
tau_bin, sf2 = [], []
for b in range(1, len(bins)):
    in_bin = idx == b
    if in_bin.sum() > 20:
        tau_bin.append(np.sqrt(bins[b-1] * bins[b]))
        sf2.append(dm2[in_bin].mean())
tau_bin, sf2 = np.array(tau_bin), np.array(sf2)

# expectation: SF^2 = 2 sigma^2 [1 - rho(tau)] + 2 sigma_n^2, with rho = exp(-tau/tau_d)
tt = np.logspace(-1.3, np.log10(3650), 300)
sf2_model = 2*sigma**2*(1 - np.exp(-tt/tau)) + 2*sigma_n**2

fig, ax = plt.subplots(1, 2, figsize=(12, 4.3), gridspec_kw={"width_ratios": [1.5, 1]})
ax[0].errorbar(t, y, yerr=sigma_n, fmt="o", ms=2.5, color="#2a6f97", elinewidth=0.6)
ax[0].set_xlabel("time (d)")
ax[0].set_ylabel(r"$\Delta m$ (mag)")
ax[0].set_title(r"DRW ($\tau = 100$ d), seasonal sampling", loc="left", fontsize=12)

ax[1].loglog(tau_bin, np.sqrt(sf2), "o", color="#2a6f97", label="binned data")
ax[1].loglog(tt, np.sqrt(sf2_model), color="#c8553d", lw=2, label="expectation")
ax[1].axhline(np.sqrt(2)*sigma_n, ls=":", color="0.4")
ax[1].axhline(np.sqrt(2*sigma**2 + 2*sigma_n**2), ls="--", color="0.4")
ax[1].text(0.06, np.sqrt(2)*sigma_n*1.12, "noise floor", color="0.3", fontsize=10)
ax[1].text(0.06, np.sqrt(2*sigma**2 + 2*sigma_n**2)*1.08, "plateau", color="0.3", fontsize=10)
ax[1].set_xlabel(r"$\tau$ (d)")
ax[1].set_ylabel(r"$SF(\tau)$ (mag)")
ax[1].set_ylim(0.04, 0.6)
ax[1].legend(frameon=False, loc="lower right", fontsize=10)
fig.tight_layout()
fig.savefig("images/structure_function.png", dpi=150)
```
